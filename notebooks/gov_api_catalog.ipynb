# Gov API Catalog Graph

This notebook is a lightweight trigger for `app/graphs/catalog.py`. Run it to compile and invoke the graph.


```python
import sys
from pathlib import Path

project_root = Path.cwd().resolve()
searched = set()
while not (project_root / "app").exists():
    if project_root in searched or project_root.parent == project_root:
        raise RuntimeError("Unable to locate project root containing the app package; open the notebook inside civic-app.")
    searched.add(project_root)
    project_root = project_root.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
print(f"Using project root: {project_root}")

from app.graphs.catalog import create_catalog_graph

graph = create_catalog_graph()
out = graph.invoke({})
out
```
