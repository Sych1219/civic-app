```python
import sys
from pathlib import Path

project_root = Path.cwd().resolve()
searched = set()
while not (project_root / 'app').exists():
    if project_root in searched or project_root.parent == project_root:
        raise RuntimeError('Unable to locate project root containing the app package; open the notebook inside civic-app.')
    searched.add(project_root)
    project_root = project_root.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
print(f'Using project root: {project_root}')

from app import create_gov_api_graph
from dotenv import load_dotenv
# https://data.gov.sg/datasets?topics=environment&resultId=d_3f9e064e25005b0e42969944ccaf2e7a
load_dotenv() 
graph = create_gov_api_graph()
result = graph.invoke(
    {
        # "request_url": "https://data.gov.sg/datasets?topics=environment&resultId=1459",
        "request_url": "https://data.gov.sg/datasets?topics=environment&resultId=d_3f9e064e25005b0e42969944ccaf2e7a",
        "auto_register": False,  # review before submitting
    },
    config={
        "run_name": "Gov API Draft",
        "dry_run": True,
    },
)
print(result.get("contract"))
print(result.get("validation_errors"))
```
