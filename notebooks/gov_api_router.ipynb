# Gov API Router Graph

This notebook routes chat-like input through the router graph to either register an API or invoke an existing one (catalog + trigger). Update the sample state as needed.


```python
import sys
from pathlib import Path
from dotenv import load_dotenv

load_dotenv()

project_root = Path.cwd().resolve()
searched = set()
while not (project_root / "app").exists():
    if project_root in searched or project_root.parent == project_root:
        raise RuntimeError("Unable to locate project root containing the app package; open the notebook inside civic-app.")
    searched.add(project_root)
    project_root = project_root.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
print(f"Using project root: {project_root}")

from app.graphs.router import create_router_graph

graph = create_router_graph()
# Example: asking for weather should route to catalog+trigger path
state = {
    "source_text": "How is the weather like Singapore in 2026-Jan-13 ?",
    # Optional explicit intent override: "register" or "invoke"
    # "metadata": {"intent": "invoke"},
}
# Set dry_run=False to actually call registry endpoints during trigger
out = graph.invoke(state, config={"dry_run": True})
out
```
